```python
import pandas as pd
df = pd.read_csv('/content/Expt11.csv')
print(df)
```

```python
display(df['Grade'].value_counts())
```

```python
display(df['Gender'].value_counts())
```

```python
display(df['Department'].value_counts())
```

```python
# Percentage distributio of grade
print(df['Grade'].value_counts(normalize=True))
```

```python
# Gender vs brade analysis
print(pd.crosstab(df['Gender'],df['Grade']))
```

```python
# Department VS Gender
print(pd.crosstab(df['Department'],df['Gender']))
```

```python
print(pd.crosstab(df['Department'],df['Grade']))
```

```python
print(pd.crosstab(df['Department'],df['Grade'],normalize='index')*100)
```

```python
# Grouping Data
print(df.groupby('Department')['Grade'].value_counts())
```

```python
data = {
'Order_ID':['O1','O2','O3','O4','O5','O6','O7','O8','O9','O10'],

'Category':['Electronics','Clothing','Electronics','Grocery','Clothing',
            'Electronics','Grocery','Clothing','Electronics','Grocery'],

'Payment_Method':['UPI','Card','UPI','Cash','Card',
                  'UPI','Cash','Card','UPI','Cash'],

'Delivery_Type':['Express','Standard','Express','Standard','Express',
                 'Standard','Express','Standard','Express','Standard'],

'Customer_Type':['New','Returning','Returning','New','Returning',
                 'New','Returning','New','Returning','New']
}
print(data)
```

```python
df_new = pd.DataFrame(data)
display(df_new['Category'].value_counts())
```

```python
display(df_new['Payment_Method'].value_counts())
```

```python
print(df_new['Category'].value_counts(normalize=True)*100)
```

```python
# Filtering Data
# Example : Order from electronics category
electronics_orders = df_new[df_new['Category'] == 'Electronics']
print(electronics_orders)
```

```python
# Sorting Data
sorted_data = df_new.sort_values(by='Category')
print(sorted_data)
```

```python
# Display all unique values
print(df_new['Category'].unique())
```

```python
# Number of Unique Value
print(df_new['Category'].nunique())
```

```python
# Crosstabulation - Category vs payment method
print(pd.crosstab(df_new['Category'],df_new['Payment_Method']))
```

```python
# Grouping
# Orders by category and payment method
print(df_new.groupby('Category')['Payment_Method'].value_counts())
```
